```python
import plotly.express as pt
import pandas as pd
import numpy as np
```

```python
comp_A=[400,500,510,520,530,540,
        550,570,580,590,600,560,
        610,620,630,640,650,660
        ]
comp_B=[300,320,340,370,390,400,
        380,400,390,430,440,500,
        480,540,600,850,1200,1220]
comp_C=[50,100,250,700,750,800,
        740,750,760,780,790,810,
        800,810,820,830,840,900]
df=pd.DataFrame({
    "A":comp_A,
    "B":comp_B,
    "C":comp_C
})
df
```

```python
print("Mean:")
print(df.mean())

print("\nMedian:")
print(df.median())

print("\nMode:")
print(df.mode().iloc[0])
```

```python
Q1=df.quantile(0.25)
Q2=df.quantile(0.50)
Q3=df.quantile(0.75)
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
Iqr = Q3-Q1
print("IQR:", Iqr)
```

```python
lower=Q1-1.5*Iqr
upper=Q3+1.5*Iqr
```

```python
for col in df.columns:
    outliers=df[(df[col]<lower[col]) | (df[col]>upper[col])][col]
    print("Outliers:", list(outliers))
```
