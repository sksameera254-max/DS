```python
import pandas as pd
df=pd.read_csv(r"C:\Users\sameera\Downloads\iris.csv")
print(df.head())
print(df)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
plt.figure(figsize=(6,3),facecolor="PINK") 
plt.bar(df["SepalLengthCm"],df["SepalWidthCm"])
plt.title("Length Graph")
plt.xlabel("Sepallenght")
plt.ylabel("SepalWidth")
plt.show()
```

```python
plt.figure(figsize=(6,3),facecolor="magenta")
plt.scatter( x=['SepalLengthCm'],y=['PetalLengthCm'],s=30,c="green",marker='*',lw=2,alpha=0.7)
sns.scatterplot(
    x='SepalLengthCm',
    y='PetalLengthCm',
    hue='Species',
    data=df
)

plt.title("Scatter Plot ")
plt.grid(True,alpha=0.8)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.kdeplot(data=df, x="SepalLengthCm", label="SepalLengthCm", fill=True)
sns.kdeplot(data=df, x="PetalLengthCm", label="PetalLengthCm", fill=True)
sns.kdeplot(data=df, x="SepalWidthCm", label="SepalWidthCm", fill=True)
sns.kdeplot(data=df, x="PetalWidthCm", label="PetalWidthCm", fill=True)
plt.title("KDE Plot ")
plt.xlabel("Values")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python

plt.figure(figsize=(6,4),facecolor="gold")
x=df["Id"]
y=df["SepalLengthCm"]
plt.plot(x,y,marker=".",markersize=3,ls="--")
plt.title("Line Graph ")
plt.xlabel("Id")
plt.ylabel("Length")
plt.grid(True,alpha=0.2)
plt.show()
```

```python
plt.figure(figsize=(6,4),facecolor="cyan")
plt.hist(df['PetalLengthCm'], bins=15, color='blue', edgecolor='black')
plt.xlabel('Length')
plt.ylabel('Frequency')
plt.title('Histogram Graph')
plt.show()
```
