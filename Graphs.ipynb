```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df={"Studs":["Raj","Rahul","Karu","Vennela","Pallavi"],
    "Marks":[86,72,91,63,78],
    "Age":[20,21,22,23,24]
    }
df=pd.DataFrame(df)
df
```

```python
plt.figure(figsize=(6,3),facecolor="PINK") 
plt.bar(df["Studs"],df["Marks"])
plt.title("Data Graph")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df2={
    "Height":[5.2,4.2,6.3,4.4,5.7,6.5,4.9,5.5],
    "Weight":[50,55,65,70,85,60,68,75]
}
df2=pd.DataFrame(df2)
df2
```

```python
plt.figure(figsize=(6,3),facecolor="pink")
plt.scatter(df2["Height"],df2["Weight"],s=30,c="green",marker='*',lw=2,alpha=0.7)
plt.title("Scatter Plot ")
plt.grid(True,alpha=0.8)
plt.show()
```

```python
plt.figure(figsize=(6,4),facecolor="magenta")
plt.scatter(df2["Height"],df2["Weight"],s=50,c="black",marker='^',lw=2,alpha=0.7)
plt.title("Scatter Plot ")
plt.grid(True,alpha=0.8)
plt.show()
```

```python
months=["jan","feb","mar","apr","may"]
sales=[2000,2300,1600,3000,1000]

plt.figure(figsize=(6,3),facecolor="pink")

plt.plot(months,sales,marker=".",markersize=10,label="Sales Data ",lw=1,ls="--")
plt.title("Data Viz ")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True,alpha=0.2)
plt.show()
```
